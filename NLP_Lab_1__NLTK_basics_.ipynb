```python
fdist = FreqDist(text3)
print(fdist)

hapaxes = fdist.hapaxes()
less_than_5 = [w for w in fdist if fdist[w] < 5]

print(f"Кількість гапаксів у тексті: {len(hapaxes)}.")
print(f"Приклади гапаксів у тексті: {hapaxes[:10]}")
print(f"Кількість слів, що вживаються менше 5 разів: {len(less_than_5)}.")
print(f"Приклади слів, що зустрічаються менше 5 разів: {less_than_5[:10]}")
```

```python
import matplotlib.pyplot as plt

word_lengths = [len(w) for w in text3 if w.isalpha()] 
length_dist = FreqDist(word_lengths)

for length in sorted(length_dist):
    print(f"Довжина {length} букв: {length_dist[length]} слів")

plt.figure(figsize=(10, 5))
length_dist.plot(title="Залежність частоти вживання слів від їхньої довжини")
plt.show()
```

```python
from datasets import load_dataset
import pandas as pd

dataset = load_dataset("momentum-lab/news-categories")

df = pd.DataFrame(dataset["train"])
df.head()
```

```python
category = "Technology & Science"

tech_df = df[df["topic"] == category].copy()
print("Кількість новин:", len(tech_df))
```

```python
keywords = [
    "amazon",
    "million",
    "support",
    "iPhone",
    "FBI",
    "war",
    "ai",
    "data",
    "Nvidia",
    "TikTok"
]
```

```python
tech_df["title_lower"] = tech_df["title"].str.lower()

for word in keywords:
    tech_df[word] = tech_df["title_lower"].str.count(word)
```

```python
tech_df["date"] = pd.to_datetime(tech_df["date"])
tech_df["date_only"] = tech_df["date"].dt.date
```

```python
daily_counts = tech_df.groupby("date_only")["data"].sum()

plt.figure()
plt.plot(daily_counts.index, daily_counts.values)
plt.xticks(rotation=90)
plt.grid(True)
plt.title("Частота використання слова 'data' за датою")
plt.xlabel("Дата")
plt.ylabel("Кількість згадувань")
plt.show()
```

```python
source_counts = tech_df.groupby("source")["data"].sum()

plt.figure()
plt.bar(source_counts.index, source_counts.values)
plt.xticks(rotation=90)
plt.grid(axis='y', linestyle='-', alpha=0.6)  
plt.title("Частота використання слова 'data' за джерелами")
plt.xlabel("Джерело")
plt.ylabel("Кількість згадувань")
```
